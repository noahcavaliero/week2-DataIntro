### 4. Who might be missing?

**Answer:** [The 100,000-copy threshold excludes all smaller indie games, niche titles, and games that sold fewer copies . The dataset was scraped circa 2016, so games released after that date are entirely absent. VGChartz itself acknowledges that its data ceased covering software sales after 2018 due to digital market share making retail estimates unreliable . Digital-only sales (Steam, eShop, PSN, Xbox Live) were not tracked by VGChartz during this period, meaning games with primarily digital distribution are underrepresented or absent .

Reasonable interpretation:
Entire categories of games are missing: free-to-play titles, mobile games, digital-only indie games, and games from regions outside the tracked retail markets (Latin America, Africa, parts of Asia) are poorly represented.

Unknown information to investigate:
The exact proportion of missing data is unknown. VGChartz claimed 3-5% retail coverage , but the representativeness of this sample for the broader market is questionable.]
